# AI-Powered Cybersecurity Threat Detection System
## Exploratory Data Analysis (EDA) & Model Prototyping

This notebook demonstrates exploratory analysis, cybersecurity feature engineering, and threat pattern discovery on network telemetry data.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="darkgrid")
df = pd.read_csv("../data/raw_network_traffic.csv")
print(f"Dataset Shape: {df.shape}")
df.head()
```

### 1. Threat Class Distribution
Analyzing the frequency and proportions of benign traffic vs malicious cyber attacks.

```python
plt.figure(figsize=(8, 4))
sns.countplot(data=df, y="label", palette="viridis")
plt.title("Distribution of Threat Categories in Network Telemetry")
plt.show()
```

### 2. Feature Analysis: Bytes and Packet Rates Across Attack Classes

```python
df['byte_rate'] = (df['src_bytes'] + df['dst_bytes']) / (df['duration'] + 1e-5)
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x="label", y=np.log1p(df["byte_rate"]), palette="magma")
plt.title("Log-Scaled Byte Rate by Attack Class")
plt.xticks(rotation=20)
plt.show()
```

### 3. SSH Brute Force Signature: Failed Logins
Examining failed authentication spikes targeting port 22.

```python
print("Failed Logins Summary by Threat Type:")
print(df.groupby("label")["failed_logins"].describe())
```
